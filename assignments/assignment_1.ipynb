```python
import matplotlib.pyplot as plt
import numpy as np


def u(z: float | np.ndarray, ug: float = 10, delta: float = 1000) -> float | np.ndarray:
    return ug *  (1 - np.exp(-z / delta) * np.cos(z / delta))


def v(z: float | np.ndarray, ug: float = 10, delta: float = 1000) -> float | np.ndarray:
    return ug *  np.exp(-z / delta) * np.sin(z / delta)


z = np.linspace(0, 4_500, 10_001)
delta = 1000
ug = 10

u = u(z, delta=delta)
v = v(z, delta=delta)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(u, z, label='u(z)')
ax1.plot(v, z, label='v(z)')
ax1.set_xlabel(r'Velocity Component / ms$^{-1}$')
ax1.set_ylabel('Height / km')
ax1.set_ylim(0, None)
ax1.set_yticks(np.arange(0, 5000, 1000))
ax1.set_yticklabels(np.arange(0, 5, 1))
ax1.grid(True, linestyle=':', alpha=0.5)
ax1.legend()

ax2.plot(u/ug, v/ug, label='Hodograph')
ax2.set_xlabel(r'u/u$_g$')
ax2.set_ylabel(r'v/v$_g$')
ax2.set_xlim(0, None)
ax2.grid(True, linestyle=':', alpha=0.5)
ax2.legend()

fig.suptitle("Mean wind profile for a boundary layer of 1 km.")

plt.show()
```
